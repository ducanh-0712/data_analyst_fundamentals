```python
def convert_to_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)
    
df['job_skills'] = df['job_skills'].apply(lambda skill_list: ast.literal_eval(skill_list) if pd.notna(skill_list) else skill_list)
```

```python
type(df['job_skills'][1])
```

```python
def special_projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return row['salary_year_avg']*1.05
    else:
        return row['salary_year_avg']*1.03
    
df['salary_year_inflated'] = df[pd.notna(df['salary_year_avg'])].apply(special_projected_salary, axis = 1)

df[pd.notna(df['salary_year_avg'])][['job_title_short','salary_year_avg', 'salary_year_inflated']]
```

```python
df_copy = df.copy()
df_copy = df_copy.explode('job_skills')
df_copy
```

```python
df_copy.groupby('job_skills').size().sort_values(ascending= True).tail(10).plot(kind='barh')
```

```python
skills_count = df_copy.groupby(['job_title_short', 'job_skills']).size().reset_index(name = 'skill_count')
skills_count[skills_count['job_title_short'] == 'Data Engineer'].sort_values('skill_count', ascending=True).tail(10).plot(kind= 'barh', x= 'job_skills', y= 'skill_count')
```

```python

skill_month = df_copy[df_copy['job_title_short'] == 'Data Analyst'].groupby(['job_posted_month','job_skills']).size().reset_index(name= 'skill_count')
```

```python
skill_month_pivot = skill_month.pivot_table(values= 'skill_count', index= 'job_posted_month', columns='job_skills')



top_skills = skill_month_pivot.sum(numeric_only= True).sort_values(ascending= False).head().index.to_list()
skill_month_pivot = skill_month_pivot.reset_index()
skill_month_pivot['month_no'] = pd.to_datetime(skill_month_pivot['job_posted_month'], format= '%b').dt.month
skill_month_pivot.sort_values(by= 'month_no', ascending=True, inplace= True)

skill_month_pivot.set_index('job_posted_month')[top_skills].plot(
    kind= 'line',
    linewidth= 3,
    linestyle= ':',
    colormap= 'plasma',
    marker= 'o',
    markersize= 5,
    figsize= (10,5)
)
plt.title('Data Jobs Skills Trend in 2023')
plt.xlabel('Month')
plt.ylabel('Job Postings')
plt.legend().set_title('Job Skills')
```
